## Úloha 2 — Test výjimky (5 b)

Funkce `deleni(a, b)` vrací podíl, při dělení nulou vyvolá `ValueError`. Napište třídu `TestDeleni` a ověřte: běžné dělení, dělení se záporným číslem, dělení nuly (`0 / 5`), dělení nulou a také to, že funkce vrací správný výsledek pro desetinné číslo (`1 / 3`).

<details>
<summary>🔎 Očekávaný výstup</summary>

```
Ran 5 tests in 0.001s

OK
```

</details>

````python
import unittest


def deleni(a, b):
    if b == 0:
        raise ValueError("Nelze dělit nulou!")
    return a / b


# TODO: Úloha 2
class TestDeleni(unittest.TestCase):
    def test_deleni_normalni(self):
        self.assertEqual(deleni(5, 2), 2.5)

    def test_deleni_zaporne(self):
        self.assertEqual(deleni(5, -1), )

    def test_deleni_nuly(self):
        self.assertEqual(deleni(0, 5), 0)

    def test_deleni_s_nulou(self):
        with self.assertRaises(ValueError):
            deleni(5, 0)

    def test_deleni_desetinne(self):
        self.assertAlmostEqual(deleni(1, 3), 0.3)

   

suite = unittest.TestLoader().loadTestsFromTestCase(TestDeleni)
runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite)
````

Funkce `omez_do_rozsahu(x, minimum, maximum)` omezí číslo do daného intervalu. Napište třídu `TestOmezDoRozsahu`, která pokryje: hodnotu uvnitř intervalu, hodnotu pod minimem, hodnotu nad maximem a obě hranice intervalu. <details>
<summary>🔎 Očekávaný výstup</summary>


````python
import unittest


def omez_do_rozsahu(x, minimum, maximum):
    return max(minimum, min(x, maximum))


# TODO: Úloha 3
class TestOmezDoRozsahu(unittest.TestCase):
    def test_hodnota_uvnitr(self):
        self.assertEqual(omez_do_rozsahu(5, 1, 10), 5)

    def test_hodnota_pod_minimum(self):
        self.assertEqual(omez_do_rozsahu(-1, 1, 10), 1)
        self.assertEqual(omez_do_rozsahu(-20, 1, 10), 1)

    def test_hodnota_nad_maximum2(self):
        self.assertEqual(omez_do_rozsahu(60, 1, 10), 10)
        self.assertEqual(omez_do_rozsahu(15, 1, 10), 10)

    def test_hodnota_na_hranici_min(self):
        self.assertEqual(omez_do_rozsahu(1, 1, 10), 1)

    def test_hodnota_na_hranici_max(self):
        self.assertEqual(omez_do_rozsahu(10, 1, 10), 10)

suite = unittest.TestLoader().loadTestsFromTestCase(TestOmezDoRozsahu)
runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite)
````

## Úloha 5 — Seznamy a výjimky (5 b)

Funkce `soucty_prvku(a, b)` sečte odpovídající prvky dvou seznamů stejné délky, jinak vyvolá `ValueError`. Napište třídu `TestSouctyPrvku` a ověřte: běžné sečtení, prázdné seznamy, seznamy se zápornými čísly, seznamy různé délky a také to, že funkce **nezmění** původní seznamy.

<details>
<summary>🔎 Očekávaný výstup</summary>

```
Ran 5 tests in 0.001s

OK
```

</details>

````python
import unittest


def soucty_prvku(a, b):
    if len(a) != len(b):
        raise ValueError("Seznamy musí mít stejnou délku!")
    return [x + y for x, y in zip(a, b)]


# TODO: Úloha 5
class TestSouctyPrvku(unittest.TestCase):
    def test_spravne_scitani(self):
        self.assertEqual(soucty_prvku([1, 2, 3], [4, 5, 6]), [5, 7, 9])

    def test_prazdne_seznamy(self):
        self.assertEqual(soucty_prvku([], []), [])

    def test_zaporna(self):
        self.assertEqual(soucty_prvku([-1, -2, -3], [-4, -5, -6]), [-5, -7, -9])

    def test_ruzna_delka(self):
        with self.assertRaises(ValueError):
            soucty_prvku([1, 2], [1, 2, 3])

    def test_nezmeni_puvodni(self):
        a1 = [1, 2, 3]
        b1 = [4, 5, 6]
        soucty_prvku(a1, b1)
        self.assertEqual(a1, [1, 2, 3])


suite = unittest.TestLoader().loadTestsFromTestCase(TestSouctyPrvku)
runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite)
````

<details>
<summary>🔎 Očekávaný výstup</summary>

```
Ran 5 tests in 0.001s

OK
```

</details>

````python
import unittest


class Zasobnik:
    def __init__(self, kapacita=3):
        self.kapacita = kapacita
        self.data = []

    def vloz(self, x):
        if len(self.data) >= self.kapacita:
            raise OverflowError("Zásobník je plný!")
        self.data.append(x)

    def vyjmi(self):
        if not self.data:
            raise IndexError("Zásobník je prázdný!")
        return self.data.pop()

    def je_prazdny(self):
        return len(self.data) == 0

    def velikost(self):
        return len(self.data)


# TODO: Úloha 6
class TestZasobnik(unittest.TestCase):
    def test_novy_zasobnik(self):
        self.assertEqual(Zasobnik().data, [])

    def test_zmena_velikosti(self):
        zasobnik = Zasobnik()
        size1 = zasobnik.velikost()
        zasobnik.vloz(1)
        self.assertNotEqual(size1, zasobnik.velikost())

    #TODO: Vyjmutí vrací poslední vložený prvek

    def test_vyjmuti_prazdny(self):
        zasobnik = Zasobnik()
        with self.assertRaises(IndexError):
            zasobnik.vyjmi()

    def test_vlozeni_do_plneho(self):
        zasobnik = Zasobnik()
        zasobnik.vloz(27)
        zasobnik.vloz(27)
        zasobnik.vloz(27)
        with self.assertRaises(OverflowError):
            zasobnik.vloz(27)

        
suite = unittest.TestLoader().loadTestsFromTestCase(TestZasobnik)
runner = unittest.TextTestRunner(verbosity=2)
runner.run(suite)
````
